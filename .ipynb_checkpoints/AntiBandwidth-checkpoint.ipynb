# Anti-Bandwidth Problem

## I. Khái niệm
- Bài toán Anti-bandwidth (ABP) là một bài toán tối ưu hóa tổ hợp trên đồ thị, được xác định là một bài toán `NP-Hard`
-  Cho một đồ thị $G=(V,E)$ với $V$ là tập các đỉnh và $E$ là tập các cạnh. Một cách đánh nhãn $f$ là một `song ánh` từ tập các đỉnh $V$ vào tập các số nguyên $\{1,...,n\}$, trong đó mỗi đỉnh nhận một nhãn duy nhất
-  Mục tiêu của bài toán Anti-bandwidth là tìm một cách đánh nhãn sao cho `khoảng cách nhỏ nhất giữa nhãn của hai đỉnh kề nhau là lớn nhất` có thể

## II. Các ràng buộc

#### 1. Định nghĩa biến số
- $X_{i, l}  = True$ $\leftrightarrow f(i) = l$ \
  nghĩa là:
  $X_{i, l}  = True$ khi và chỉ khi đỉnh $i$ được gán nhãn $l$

#### 2. Các ràng buộc

- Ràng buộc đỉnh (Vertices)
    - Ý nghĩa: Đảm bảo rằng mỗi nhãn $l$ chỉ được gấn cho `một đỉnh duy nhất` trong đồ thị
    - Công thức: $$\bigwedge_{l \in \{1, \dots, |V|\}} \sum_{i \in V} X_{i, l} = 1$$
    

- Ràng buộc nhãn (Labels)
    - Ý nghĩa: Đảm bảo rằng mỗi đỉnh $i$ chỉ nhận `duy nhất một nhãn` trong đồ thị
    - Công thức: $$\bigwedge_{i \in V}\ \sum_{l \in \{1, \dots, |V|\}} X_{i,l} = 1$$

- Ràng buộc mục tiêu (OBJ-k)
    - Ý nghĩa: Hiệu số giữa nhãn của đỉnh $i$ và đỉnh $i'$ không được $\le K$. 
    - Công thức: $$\bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} (X_{i, l'} + X_{i', l'}) \le 1$$
      > Trong bất kỳ nhóm $k+1$ nhãn liên tiếp nào, chỉ có tối đa một nhãn được phép gán nhãn $l'$ hoặc là đỉnh $i$, hoặc là đỉnh $i'$. Nếu cả hai đỉnh kề nhau này đều "rơi" vào cùng một cửa sổ nhãn có độ rộng $k+1$, thì khoảng cách giữa chúng chắc chắn sẽ $\le k$, vi phạm điều kiện bài toán.
    - Dạng khai triển:\
      $$(X_{i, 1} + X_{i', 1}) + (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) \le 1  $$
      $$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 2 + k - 1} + X_{i', 2 + k - 1}) +  (X_{i, 2 + k} + X_{i', 2 + k})\le 1$$ 
      $$\dots$$
      
      > Hiểu đơn giản là không thể gán nhãn cho 2 đỉnh liên tiếp mà hiệu của hai nhãn đó quá gần nhau

#### 3. Encoding OBJ-k

$$(X_{i, 1} + X_{i', 1}) + (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) \le 1  $$
$$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) +  (X_{i, 2 + k} + X_{i', 2 + k})\le 1$$ 
      $$\dots$$

- Cách 1: Tách riêng $X_{i,l_2}$ và $X_{i', l_2}$ để encode
    - Để tách ra làm hai ta cần: thêm ràng buộc ta cần loại trừ bỏ trường hợp cả 2 tổng sau khi tách cùng bằng 1, vì nếu cả 2 tổng sau khi tách cùng bằng 1 thì tổng ban đầu sẽ bằng 2 nghĩa là $\ge 1$
    - Nghĩa là:
      $$ \neg (\bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} X_{i, l'} = 1 \ \ \ \land \  \bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} X_{i', l'} = 1) $$
      $$ \equiv \bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} X_{i, l'} < 0 \ \ \ \lor \  \bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} X_{i', l'} < 0$$
    - nên ta sẽ có công thức đầy đủ là
  $$\equiv \bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} X_{i, l'} \le 1 \ \ \ \land \  \bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} X_{i', l'} \le 1 \ \ \ \land (\bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} X_{i, l'} \le 0 \ \ \ \lor \  \bigwedge_{(i,i') \in E} \bigwedge_{l_2 = 1} ^ {l_2 = |V| - k} \ \sum_{l' = l_2} ^ {l_2 + k} X_{i', l'} \le 0)$$
    - Đơn giản hóa
  $$AMO(v_1, k) \land AMO(v_2, k) \land [AMZ(v_1, k) \lor AMZ(v_2, k)]$$
  Lưu ý: $v$ là tập hợp tất cả các biến tương ứng
      

- Cách 2: Không tách riêng mà quy về dạng mặc định để encode
$$(X_{i, 1} + X_{i', 1}) + (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) \le 1  $$
$$\ \ \ \ \ \ \  \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \underline{(X_{i', 1}) + (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) +  (X_{i, 2 + k}) \le 1}  $$
$$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) +  (X_{i, 2 + k} + X_{i', 2 + k})\le 1$$ 
      $$\dots$$
    - **suy ra**
$$(X_{i, 1} + X_{i', 1}) + (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) \le 1  $$
$$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) +  (X_{i, 2 + k} + X_{i', 2 + k})\le 1$$ 
      $$\dots$$
    - **Tuy nhiên:** chiều ngược lại thì không đúng
    > Nếu không yêu cầu quá khắt khe thì ta có thể encode `Ladder ở phía trên` luôn. Vì nếu `thỏa mãn Ladder ở phía trên thì sẽ thỏa mãn ladder ở phía dưới`


- Trường hợp ta làm theo cách 2 và ta không muốn bỏ bất kì trường hợp nào cả thì ta làm thế nào?
    - Ta phải trả lời câu hỏi nếu encode Ladder ở phía trên thì ta thiếu mất trường hợp nào?
        - Hãy để ý kích thước của các `labels` liên tiếp. Ở các `biểu thức trung gian` (là phần được gạch chân) kích thước của các `labels` liên tiếp không còn là $k$ mà là $k + 1$. Nhìn nào số hạng đầu tiên và số hạng cuối cùng của biểu thức trung gian là ta sẽ thấy điều đó
        - Và ta không cấm hiệu của 2 đỉnh liền kề $= k + 1$ nhưng Ladder ở phía trên vứt bỏ trường hợp này vì điều kiên $\le 1$
    - Phương án:
        - Nếu ta không đảm bảo các biểu thức trung gian sẽ $\le 1$ thì trong quá trình nối các `Block` ta sẽ không nối các biểu thức trung gian nghĩa là bỏ qua nó
$$(X_{i, 1} + X_{i', 1}) + (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) \le 1  $$
$$\ \ \ \ \ \ \  \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \underline{(X_{i', 1}) + (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) +  (X_{i, 2 + k}) }  $$
$$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ (X_{i, 2} + X_{i', 2}) + \dots + (X_{i, 1 + k} + X_{i', 1 + k}) +  (X_{i, 2 + k} + X_{i', 2 + k})\le 1$$ 
      $$\dots$$
> Việc này khiến code phức tạp hơn nhưng đổi lại là số mệnh đề giảm vì không phải nối tất cả các dòng, và có thể phủ hết các trường hợp
