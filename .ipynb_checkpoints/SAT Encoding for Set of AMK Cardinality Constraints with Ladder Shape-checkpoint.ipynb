# SAT Encoding for Set of At-Most-K Cardinality Constraints with Ladder Shape

## I. Khái niệm
**1. Định nghĩa:**
- Cho một tập hợp các biến boolean $\Omega = \{X_1, X_2, ..., X_3\}$ và một số $w$ thỏa mãn $1 < w \le n$, một hình dạng `thang` AMK với chiều rộng là $w$ được xây dựng theo công thức sau
  $$Ladder(\Omega, w)=\bigwedge_{i=0}^{n-w}\left(\sum_{j=i+1}^{i+w} X_j \le k\right)\qquad 1 \le k < w $$

$X_1 + X_2 + X_3+ X_4 \le k$ \
$\ \ \ \ \ \ \ \ \ $ $X_2 + X_3+ X_4 + X_5 \le k$ \
$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ $ $X_3 + X_4 + X_5 + X_6 \le k$ \
$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ $ $X_4 + X_5 + X_6 + X_7 \le k$ \
$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \  \ $ $X_5 + X_6 + X_7 + X_8 \le k$ \
$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \  \ \ \ \ \ \ \ \ \ \  \ $ $X_6 + X_7 + X_8 + X_9 \le k$ \
$\ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \ \  \ \ \ \ \ \ \ \  \ \ \ \ \ \ \ \ \ \ \ \ \ $ $X_7 + X_8 + X_9 + X_{10} \le k$

**2. Các công thức encoding**

**Lưu ý:**
- Biến phụ $R_{i,j,s}$ là tổng có $j$ phần tử của $Block$ thứ $i$ `lớn hơn hoặc bằng` $s$
    + $R_{i,j,s} = 1 \leftrightarrow \sum X \ge k$
    + $R_{i,j,s} = 0 \leftrightarrow \sum X \le k - 1$

**Công thức**
- **Các công thức khẳng định**
    - $X_{i,j} \rightarrow R_{i,j,1} (1 \le j \le w - 1)$
    - $R_{i,j - 1, s} \rightarrow R_{i, j, s} (2 \le j \le w - 1, 1 \le s \le min(j - 1, k))$
       > Vì ta biết chắc chắn rằng $X_1 + X_2 + ... + X_j$ không thể nào lớn hơn $s$ nếu $j < s$ nên $1 \le s \le min(j - 1, k)$
    - $X_{i,j} \land R_{i, j - 1, s - 1} \rightarrow R_{i,j,s} (2 \le j \le w - 1, 2 \le s \le min(j,k))$
    - $\neg X_{i, j} \rightarrow \neg R_{i,j,j} (1 \le j \le k)$
- **Các công thức phủ định**
    - $\neg R_{i, j - 1, s - 1} \rightarrow \neg R_{i,j,s} (2 \le j \le w - 1, 2 \le s \le min(j, k))$
    - $\neg X_{i, j} \land \neg R_{i, j -1, s} \rightarrow \neg R_{i,j,s} (2 \le j \le w - 1, 1 \le 1 \le min(j-1, k))$
    - $X_{i,j} \rightarrow \neg R_{i, j - 1, k} (2 \le j \le w - 1, 1 \le s \le min(j-1, k))$
      > Những công thức này để đảm bảo $R$ không bị `ảo` (đọc bài `Sequential counter AMK` để biết thêm chi tiết)
- **Công thức ràng buộc**
    - $X_{i,j} \rightarrow \neg R_{i,j-1,k} (k + 1 \le j \le w)$
      > nếu đã đạt tối đa $k$ biến $True$ thì không được có thêm 1 biến $True$ nào nữa 

**3. Cách nối các Block**

**Ý tưởng**

- Bản thân trong có trình encoding từng `Block` bằng `sequential counter` thì từng tổng của `Block` đã được encoding ($X_1 + X_2 + ... + X_j \le 2)$
- Ta sẽ nối các dòng tương ứng bằng cách `loại bỏ các trường hợp` khiến cho `tổng sau khi được kết nối vượt quá k` (chính là 2 ta đang xét ở ví dụ này)


<img src= "D:\SAT\SAT_TranDucThanh_24021630\image\noiBlock2.png" width="700"/>

**Nối**
- Nối `khối` $A_1$ với $A_2$
    - $X_1 + X_2 + X_3 + X_4 \le 2$ đã đầy đủ không nếu với các gì
      
    - Nối $X_2 + X_3 + X_4 \le 2$ với $X_5 \le 2$ thành $X_1 + X_2 + X_3 + X_4 + X_5 \le 2$
      Cần loại bỏ:
        - TH1: Tổng 1 bằng 2, tổng 2 bằng 1 \
          $\neg((X_2 + X_3 + X_4 = 2) \land (X_5 = 1))$ \
          $\equiv (X_2 + X_3 + X_4 \neq 2) \lor (X_5 \neq 1)$ \
          $\equiv (X_2 + X_3 + X_4 < 2) \lor (X_5 < 1)$ \
          $\equiv \neg R_{1,3,2} \lor \neg R_{2,1,1}$
          
        - TH2: $X_2 + X_3 + X_4 = 1$ và $X_5 = 2$ (không bao giờ xảy ra)
        - TH3: $X_2 + X_3 + X_4 = 2$ và $X_5 = 2$ (không bao giờ xảy ra)
     
    - Nối $X_3 + X_4 \le 2$ với $X_5 + X_6 \le 2$ thành $X_3 + X_4 + X_5 + X_6 \le 2$ Cần loại bỏ:
 
        - TH1: Tổng 1 bằng 2, tổng 2 bằng 1 \
          $\neg((X_3 + X_4 = 2) \land (X_5 + X_5 = 1))$ \
          $\equiv (X_3 + X_4 \neq 2) \lor (X_5 + X_6 \neq 1)$ \
          $\equiv (X_3 + X_4 < 2) \lor (X_5 + X_6 < 1)$ \
          $\equiv \neg R_{1,2,2} \lor \neg R_{2,2,1}$
      
        - TH2: Tổng 1 bằng 1, tổng 2 bằng 2 \
          $\neg((X_3 + X_4 = 1) \land (X_5 + X_5 = 2))$ \
          $\equiv (X_3 + X_4 \neq 1) \lor (X_5 + X_6 \neq 2)$ \
          $\equiv (X_3 + X_4 < 1) \lor (X_5 + X_6 < 2)$ \
          $\equiv \neg R_{1,2,1} \lor \neg R_{2,2,2}$
     
        - TH3: Tổng 1 bằng 2, tổng 2 bằng 2 \
          $\neg((X_3 + X_4 = 2) \land (X_5 + X_5 = 2))$ \
          $\equiv (X_3 + X_4 \neq 2) \lor (X_5 + X_6 \neq 2)$ \
          $\equiv (X_3 + X_4 < 2) \lor (X_5 + X_6 < 2)$ \
          $\equiv \neg R_{1,2,2} \lor \neg R_{2,2,2}$
     
    - 2 dòng tiếp theo tương tự nối 2 dòng đầu
 
- Nối các `Block` khác tương tự

**Công thức tổng quát**

$$
\bigwedge_{j=2}^{w}
\bigwedge_{p=1}^{k}
\left(
\sum_{q=1}^{w-j+1} x_{i,q} \le k-p
\;\vee\;
\sum_{q'=1}^{j-1} x_{i+1,q'} \le p-1
\right)
$$

- Lưu ý:
    - ở công thức này $q$ chỉ số thứ tự chứ không chỉ chỉ số của biến (không phải chỉ $X_1, X_2$ mà là chỉ $X$ ở ví trí thức nhất trong tổng, $X$ ở ví trí thứ 2 trong tổng
)


- Ý nghĩa của tham số
    - $i$ số thứ tự của `Block`
    - $w$ kích thước của `Block`
    - $k$ là giới hạnh AMK
    - $j$ là vị trí chia của sổ giữa 2 `Block`
    - $p$ là số biến $True$ mà `Block` sau dùng (dùng để phân phối biến $True$)
      


- Ý tưởng: Ta cần đảm bảo tổng sau khi kết hợp không vượt quá $k$ \
    $\neg(\sum X_{i,q} + \sum X_{i,q'} \ge k + 1)$ \
    $\equiv\neg((\sum X_{i,q} \ge k + 1 - p) \land (\sum X_{i, q'} \ge p))$ \
    $\equiv(\sum X_{i,q} \le k - p) \lor (\sum X_{i, q'} \le p - 1)$

## III. Code

Lưu ý:
- Ở code dưới đây các chỉ số đều bắt đầu từ 0 nến lùi 1 đơn vị so với công thức

```python
from pysat.solvers import Glucose3
import math
```

**a. AMK cho từng Block**

```python
def AMK(g, block, number_block, k):
    w = len(block)
        
    for j in range(0, w - 1): 
        Xij = block[j]
        Rij1 = SCVarId(number_block, j, 0)
        g.add_clause([-Xij, Rij1])
        
   
    for j in range(1, w - 1): 
        for s in range(0, min(j, k)): 
            Rijm1s = SCVarId(number_block, j - 1, s)
            Rijs = SCVarId(number_block, j, s)
            g.add_clause([-Rijm1s, Rijs]) 

    for j in range(1, w - 1):
        for s in range(1, min(j + 1, k)):
            Xij = block[j]
            Rijm1sm1 = SCVarId(number_block, j - 1, s - 1)
            Rijs = SCVarId(number_block, j, s)
            g.add_clause([-Xij, -Rijm1sm1, Rijs])
    
    for j in range(0, k): 
        Xij = block[j]
        Rijj = SCVarId(number_block, j, j)
        g.add_clause([Xij, -Rijj]) 

    for j in range(1, w - 1):
        for s in range(1, min(j + 1, k)):
            Rijm1sm1 = SCVarId(number_block, j - 1, s - 1)
            Rijs = SCVarId(number_block, j, s)
            g.add_clause([Rijm1sm1, -Rijs]) 

    for j in range(1, w - 1):
        for s in range(0, min(j, k)):
            Xij = block[j]
            Rijm1s = SCVarId(number_block, j - 1, s)
            Rijs = SCVarId(number_block, j, s)
            g.add_clause([Xij, Rijm1s, -Rijs])

    for j in range(k, w): 
        Xij = block[j]
        Rijm1k = SCVarId(number_block, j - 1, k - 1)
        g.add_clause([-Xij, -Rijm1k])
```

**b. Xác định Area thứ number có những phần tử nào**

```python
def Area(vars, weight, number):
    
    start_index = weight * number
    end_index = min(start_index + weight, len(vars))

    return vars[start_index:end_index]
```

**c. Encoding cho từng Area**

```python
def EncodingAreas(g, area, number_area, k, n_Areas, weight):
    number_block1 = NumberBlock(number_area, True, weight)
    number_block2 = NumberBlock(number_area, False, weight)
    block1 = area
    block2 = area[::-1]
    if number_area != 0:
        AMK(g, block1, number_block1, k)
    if number_area != n_Areas - 1:
        AMK(g, block2, number_block2, k)
```

**d. Nối các Area**

```python
def ConnectAreas(g, number_area1, number_area2, k, weight):
    w = weight
    number_block1 = NumberBlock(number_area1, False, weight)
    number_block2 = NumberBlock(number_area2, True, weight)
    for j in range (2, w + 1):
        for p in range (1, k + 1):
            j1 = (w - j + 1) - 1 
            s1 = (k - p + 1) - 1
            R_var1 = SCVarId(number_block1, j1, s1)

            j2 = (j - 1) - 1
            s2 = (p - 1 + 1) - 1

            if s1 >= 0 and s2 >= 0:
                R_var2 = SCVarId(number_block2, j2, s2)
                g.add_clause([-R_var1, -R_var2])
```

**e. Sequential Constraint AMK bao gồm encoding và nối**

```python
def SC_AMK(g, vars, k, weight):
    n_var = len(vars)
    n_Areas = math.ceil(len(vars)/weight)

    for number in range (n_Areas):
        area = Area(vars, weight, number)
        EncodingAreas(g, area, number, k, n_Areas,  weight)
    for number in range (n_Areas - 1):
        ConnectAreas(g, number, number + 1, k, weight)
```

**f. Hàm để lấy id các biến phụ**

```python
def KeySCVar(i, j, s):
    return f"{i}_{j}_{s}"
```

```python
def SCVarId(i, j, s):

    key = KeySCVar(i, j, s)

    if key in dictionary_id:
        return dictionary_id[key]

    dictionary_id[key] = next_id[0]
    next_id[0] += 1

    return dictionary_id[key]
```

```python
dictionary_id = {}
next_id = [1]
```

**g. Xác định số thứ tự của 1 Block khi biết số thứ tự của Area tương ứng**

```python
def NumberBlock(number_of_area, isFirst, weight):
    number_of_block = 2 * number_of_area
    if isFirst:
        number_of_block -= 1
    return number_of_block
```

**h. Main để kiểm tra**

```python
def main():
    g = Glucose3()
    global dictionary_id, next_id
    dictionary_id = {}
    vars = list(range(1,11))
    weight = 4
    k = 2
    next_id = [max(vars) + 1]
    
    SC_AMK(g, vars, k, weight)
    
    g.add_clause([2])

    result = g.solve()
    print("SAT:", result)

    if result:

        model = g.get_model()
        true_vars = [x for x in model if x > 0]
        print("Biến nhận giá trị 1:")
        print(true_vars)

if __name__ == "__main__":
    main()
```
